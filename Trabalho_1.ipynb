### **Organizando os dados**

**x_train** e **x_test** -> armazenam as classes preditoras (características do modelo)

**y_train** e **y_test** -> variáveis rótulos, aquilo que queremos prever

As variáveis de train, dizem respeito às classes preditoras(x_train) e as classes alvo (y_train)

E as classes test, dizem respeito ao desempenho do modelo aprós o treinamento.

```python
# Aqui relacionamos todos os atributos, ignorando apenas a data e o ID a fim de prever impactos econômicos
## Valores de treino
x_train = train.loc[ : , ["Magnitude", "Fatalities", "Earthquake", "Flood", "Hurricane", "Tornado", "Wildfire", "Brazil", "China", "India", "Indonesia", "Japan", "USA"]].values
x_test = test.loc[ : , ["Magnitude", "Fatalities", "Earthquake", "Flood", "Hurricane", "Tornado", "Wildfire", "Brazil", "China", "India", "Indonesia", "Japan", "USA"]].values

#Selecionando o target
y_train = train['Economic_Loss($)'].values
y_test = test['Economic_Loss($)'].values
```

Gerando os tensores

```python
import torch
```

```python
# preparando e convertendo os dados em tensores, para facilitar calculos e vizualizações

# transformações:         df -> array NumPy | array NumPy -> tensor PyTorch
x_train = torch.tensor(x_train, dtype = torch.float)
x_test  = torch.tensor(x_test,  dtype = torch.float)
y_train = torch.tensor(y_train, dtype = torch.float) # tensor que usaremos como alvo do treinamento
y_test  = torch.tensor(y_test,  dtype = torch.float) # tensor que usaremos para comparar com as previsões do modelo
```

# Normalizando os dados de entrada

Após obtermos essas valore, precisamos **normalizar** os valores das entradas porque eles estão muito altos e posterioremente podem causar problemas. Assim, colocando todos os números no intervalo entre 0 e 1.

```python
#Maximo e minimo de cada coluna
maximo1 = x_train.max(axis = 0).values
minimo1 = x_train.min(axis = 0).values

maximo2 = x_test.max(axis = 0).values
minimo2 = x_test.min(axis = 0).values

minimo1[0] = 0
minimo2[0] = 0

##Normalizando

x_train = (x_train - minimo1) / (maximo1 - minimo1)
x_test = (x_test - minimo1) / (maximo1 - minimo1)

x_train
```

# Configurando a rede neural com funções do PyTorch.

Nossa rede neural tem 3 camadas lineares e função de ativação ReLU

*   Nossa primeira camada gera 16 neurônios
*   A segunda camada gera 8 neurônios
*   A camada oculta (self.output_layer) recebe os 8 neurônios e produz o número de saída definido em output_size
*   Então, nossa função de ativação ReLU introduz a não-linearidade





```python
import torch.nn as nn

#model = torch.nn.Sequential(
   # torch.nn.Linear(13, 16),
   # torch.nn.ReLU(),
   # torch.nn.Linear(16, 8),
  #  torch.nn.ReLU(),
  #  torch.nn.Linear(8, 1)
#)

class NeuralNewtwork(nn.Module):
  def __init__(self, input_size, output_size):
    super(NeuralNewtwork, self).__init__()
    self.input_layer = nn.Linear(input_size, 16)
    self.hidden_layer1 = nn.Linear(16, 8)
    self.output_layer = nn.Linear(8, output_size)
    self.relu = nn.ReLU()


  def forward(self, x):
    out = self.relu(self.input_layer(x)) # passa pela camada de entrada e aplica ReLU
    out = self.relu(self.hidden_layer1(out)) # passa pela camada oculta e aplica ReLU
    out = self.output_layer(out) # saída final
    return out
```

# Iniciando o modelo e passando os parametros com as dimensões do dataset

```python
input_size = 13 # variáveis preditoras
output_size = 1 # variáveis de saída
model = NeuralNewtwork(input_size, output_size)
```

```python
## Função de custo
loss_fn = torch.nn.MSELoss()
```

# Treinando com MiniBatches

```python
#learning_rate = 0.5 ##1e-6
optimizer = torch.optim.Adam(model.parameters(), lr=0.05)

#Treinando
epochs = 1000

#Configurando os minibatches
batch_size = 100
batch_no = len(x_train) // batch_size

for epoch in range(epochs):
  #Cria minibatches
  for i in range(batch_no):
    start = i * batch_size
    end = start + batch_size

    x_var = x_train[start:end]
    y_var = y_train[start:end]

    #Limpando os gradientes do ultimo uso de loss.backward()
    optimizer.zero_grad()

    #Passo forward
    output_train = model(x_var)

    #Computando a loss
    loss = loss_fn(output_train, y_var)

    if (epoch % (epochs / 10) == 0 and i == 0) :
      print(epoch, loss.item()/ 1e+18)

    #calculando os gradientes
    loss.backward()

    #Atualizando os parametros
    optimizer.step()
```
